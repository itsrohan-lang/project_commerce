```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
catalog_name = 'ecommerce'
```

```python
df = spark.table(f"{catalog_name}.silver.slv_order_items")

df.limit(10).display()
```

```python
df = df.withColumn(
    "gross_amount",
    col("quantity") * col("unit_price")
)

df = df.withColumn(
    "discount_amount",
    ceil(col("gross_amount")*col("discount_pct")/100.0)
)

df=df.withColumn(
    "sale_amount",
    col("gross_amount") - col("discount_amount") + col("tax_amount")
)

df = df.withColumn("date_id",date_format(col("dt"),"yyyyMMdd").cast(IntegerType()))

df = df.withColumn(
    "coupon_flag",
    when(col("coupon_code").isNotNull(),lit(1)).otherwise(lit(0))
)

df.limit(5).display()
```

```python
fx_rates = {
    "INR": 1.00,
    "AED": 24.18,
    "AUD": 57.55,
    "CAD": 62.93,
    "GBP": 117.98,
    "SGD": 68.18,
    "USD": 88.29,
}

rates = [(k,float(v)) for k,v in fx_rates.items()]
rates_df = spark.createDataFrame(rates, ["currency","inr_rate"])
rates_df.show()
```

```python
df = (
    df
    .join(rates_df,
          rates_df.currency == upper(trim(col("unit_price_currency"))),
          "left"
          )
    .withColumn("sale_amount_inr",col("sale_amount")* col("inr_rate"))
    .withColumn("sale_amount_inr",ceil(col("sale_amount_inr")))
)
```

```python
df.limit(5).display()
```

```python
orders_gold_df = df.select(
    col("date_id"),
    col("dt").alias("transaction_date"),
    col("order_ts").alias("transaction_ts"),
    col("order_id").alias("transaction_id"),
    col("customer_id"),
    col("item_seq").alias("seq_no"),
    col("product_id"),
    col("channel"),
    col("coupon_code"),
    col("coupon_flag"),
    col("unit_price_currency"),
    col("quantity"),
    col("unit_price"),
    col("gross_amount"),
    col("discount_pct").alias("discount_percent"),
    col("discount_amount"),
    col("tax_amount"),
    col("sale_amount").alias("net_amount"),
    col("sale_amount_inr").alias("net_amount_inr")
)
```

```python
orders_gold_df.limit(5).display()
```

```python
orders_gold_df.write.format("delta") \
    .mode("overwrite") \
    .option("mergeSchema","true") \
    .saveAsTable(f"{catalog_name}.gold.gld_fact_order_items")
```
